### Series.rank, DataFrame.rank
- method(동점 처리)
----
        - """
            s.rank(
                method='average',    # 'average' | 'min' | 'max' | 'first' | 'dense'
                ascending=True,      # True: 작은 값이 1위, False: 큰 값이 1위
                na_option='keep',    # 'keep' | 'top' | 'bottom' (NaN의 상대적 위치)
                pct=False            # True면 [0,1] 구간의 백분위 비율로 반환
            )
         - """
----
    - 'average'(기본): 동점들의 순위를 평균으로 부여 (예: 2등,3등이면 둘 다 2.5)
    - 'min': 동점 모두 가장 작은 등수(=공동 2등이면 둘 다 2)
    - 'max': 동점 모두 가장 큰 등수(=둘 다 3)
    - 'dense': 중간 등수 건너뛰지 않음 (1,2,2,3,4…)
    - 'first': 원래 등장 순서를 타이브레이커로 사용(재현성 필요할 때 유용)
- ascending: 내림차순 랭킹(값 클수록 1위)이면 ascending=False.
- pct=True: 랭크를 0~1 사이 비율로 변환(누적백분위 느낌).

```python
import pandas as pd
import numpy as np

# 1. 예제 Series 생성
# - 동점값 여러 개 포함 (20 세 개, 50 두 개)
# - NaN 값 포함
# - 음수, 양수, 큰 값 섞음
s = pd.Series([30, 10, 20, 20, 50, 20, 50, np.nan, 5, 100, np.nan])

print("=== 원본 Series ===")
print(s)
print()

# 2. 기본 rank()
# - 오름차순(ascending=True), 동점은 평균(method='average')
# - NaN은 NaN으로 유지
print("=== 기본 rank() (ascending=True, method='average') ===")
print(s.rank())
# 5→1, 10→2, 20들(3,4,5)→4.0, 30→6, 50들(7,8)→7.5, 100→9, NaN→NaN
print()

# 3. method='min'
# - 동점값에 대해 '최소 등수'를 부여
# - SQL RANK()와 유사
print("=== method='min' ===")
print(s.rank(method='min'))
# 20 세 개 모두 3, 50 두 개 모두 7
print()

# 4. method='max'
# - 동점값에 대해 '최대 등수'를 부여
print("=== method='max' ===")
print(s.rank(method='max'))
# 20 세 개 모두 5, 50 두 개 모두 8
print()

# 5. method='dense'
# - 중간 등수를 건너뛰지 않음
# - SQL의 DENSE_RANK()와 동일
print("=== method='dense' ===")
print(s.rank(method='dense'))
# 5→1, 10→2, 20(3개)→3, 30→4, 50(2개)→5, 100→6
print()

# 6. method='first'
# - 등장 순서대로 순위를 부여 (동점이라도 순서에 따라 달라짐)
print("=== method='first' ===")
print(s.rank(method='first'))
# 20 세 개라도 3,4,5 순으로 각각 다르게 등수 부여됨
print()

# 7. ascending=False
# - 내림차순 정렬 (큰 값이 1등)
print("=== ascending=False (내림차순) ===")
print(s.rank(ascending=False))
# 100→1, 50들→(2+3)/2=2.5, 30→4, 20들→6, 10→8, 5→9, NaN→NaN
print()

# 8. pct=True
# - 등수를 전체 유효 값 개수로 나눈 '백분위 비율' 반환
print("=== pct=True (백분위 비율) ===")
print(s.rank(pct=True))
# 1~9등을 9로 나눈 값 (NaN 제외)
print()

# 정리
# - method 옵션에 따라 동점 처리 방식이 달라져 값이 달라질 수 있음
# - ascending=False는 등수 순서를 뒤집음
# - pct=True는 정규화된 비율값을 반환함
# - rank() 사용할 때는 method와 ascending을 명시하는 것이 안전함
```

### Pivot
- pivot: 재배열 전용. 동일한 (index, columns) 조합에 중복이 없어야 함. 집계(aggregation) 안 함.
- pivot_table: 집계 포함 피벗. 중복이 있어도 aggfunc로 요약. 가장 범용적.
- crosstab: 범주×범주 빈도표(count) 만들기에 특화. 비율(normalize)도 간편.

- pivot
    - df.pivot(index='행으로 둘 컬럼', columns = '열로 둘 컬럼', values='셀에 채울 값')

```python
# pclass, sex 따른 fare의 평균을 둔다.
avg_fare=df.groupby(['pclass','sex'])['fare'].mean().reset_index() ## pivot을 돌리기 위해 다시 리셋 인덱스!
```
