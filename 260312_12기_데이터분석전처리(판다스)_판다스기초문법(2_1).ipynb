### 데이터 병합

- why? 병합하는 이유
    - 기업에는 모든 DB데이터가 나눠서 메타데이터 정리되어 있다.
    - 데이터 병합은 필수
- concat
    - 덩어리와 덩어리가 axis기준으로 붙여지는 것
    - 데이터프레임 + 데이터프레임 붙여지는 형태
    - pd.concat([df1,df2], axis= 1,0 둘 중 하나)
- merge
    - 공통된 컬럼이 존재
    - 병합 방법이 inner, left, right ,outer 조인연산방법
    - join
    - pd.merge(df1,df2, how='기준',on =공통컬럼)

```python
test1 =pd.DataFrame({'학회원분반번호':[1,2,3,4,5],
             '학회원분반':['Python','R','SQL','데이터전처리','데이터모델링']})

test2 =pd.DataFrame({'학회원분반번호':[1,2,3,4,6],
             '학회원수':[100,200,300,400,500]})
```

```python
## merge
pd.merge(test1,test2, how='inner',on='학회원분반번호')
```

```python
pd.merge(test1,test2, how='outer',on='학회원분반번호')
```

```python
pd.concat([test1, test2],axis =1)
```

```python
pd.concat([test1, test2],axis =0)
```

- cut() , qcut()
- 데이터를 나눈다.
- 데이터 구간화 binning 작업시 사용
    - 연속형은 고정된 간격인 bins 나누는 것
    - 수치형 데이터는 범주형 데이터로 구간화 변환   
- cut
    - 구간의 크기가 일정하게, 값의 분포는 고려하지 않는다. 구간 개수는 우리가 직접 설정이 가능하고
    - pd.cut(데이터프레임_시리즈형태, bins= 지정된 값)
- qcut
    - 분위수 Quantile 기준을 나누는 차이
    - 데이터의 분포가 개수가 거의 동일한 그룹으로 나뉘어 진다.
    - 등간격이 아니라 분포를 고려해서 구간을 자동으로 설정
    - 구간에 속하는 데이터의 개수는 동일하다.
    - qd.qcut(데이터프레임_시리즈형태, q=지정)

```python
#pd.cut(df['age'],bins=4).value_counts()

df['age_cut']=pd.cut(df['age'],bins=4)
```
